```python
# Function to display all scores with their index
def display_scores(scores):
    print("\n--- Student Scores ---")
    for i in range(len(scores)):
        print("Index", i, ":", scores[i])

# Function to search for a specific score
def search_score(scores):
    target = int(input("Enter score to search: "))
    found = False
    for i in range(len(scores)):
        if scores[i] == target:
            print("Score", target, "found at index", i)
            found = True
    if not found:
        print("Score not found.")

# Function to compute the average score
def compute_average(scores):
    total = sum(scores)
    average = total / len(scores)
    print("Average Score:", round(average, 2))

# Function to update a score using its index
def update_score(scores):
    display_scores(scores)
    index = int(input("Enter index to update: "))
    if 0 <= index < len(scores):
        new_score = int(input("Enter new score: "))
        scores[index] = new_score
        print("Score updated!")
    else:
        print("Invalid index.")

# Main program
scores = [85, 92, 78, 88, 95]

while True:
    print("\n=== Student Score Analyzer ===")
    print("1. Display Scores")
    print("2. Search Score")
    print("3. Compute Average Score")
    print("4. Update Score")
    print("5. Display Updated Scores")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        display_scores(scores)
    elif choice == "2":
        search_score(scores)
    elif choice == "3":
        compute_average(scores)
    elif choice == "4":
        update_score(scores)
    elif choice == "5":
        display_scores(scores)
    elif choice == "6":
        print("Exited program")
        break
    else:
        print("Invalid choice. Try again.")
```
